```python
import pandas as pd
import re
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer

print("Libraries imported successfully!")
```

```python

# Create a small customer review dataset

reviews = [
    "This product is very good and useful",
    "The product is bad and not worth buying",
    "I love this amazing product",
    "The quality is poor and disappointing",
    "This is a good product with great quality",
    "I hate this product because it is useless",
    "The service was excellent and very helpful",
    "The delivery was late and the service was bad"
]

# Convert the reviews into a DataFrame
df = pd.DataFrame({"Review": reviews})

# Display the dataset
print("Customer Review Dataset:")
display(df)
```

```python

print("Number of reviews:", len(df))
print("\nDataset information:")
print(df.info())
```

```python

# Convert all customer reviews to lowercase

df["Lowercase_Review"] = df["Review"].str.lower()

# Display original and lowercase text
display(df[["Review", "Lowercase_Review"]])
```

```python

# Tokenization: split each review into individual words

df["Tokens"] = df["Lowercase_Review"].apply(word_tokenize)

# Display the lowercase reviews and their tokens
display(df[["Lowercase_Review", "Tokens"]])
```

```python

# Get English stopwords
stop_words = set(stopwords.words("english"))

# Remove stopwords from each review
df["Cleaned_Tokens"] = df["Tokens"].apply(
    lambda tokens: [word for word in tokens if word.isalpha() and word not in stop_words]
)

# Display the results
display(df[["Tokens", "Cleaned_Tokens"]])
```

```python

# Convert cleaned token lists back into sentences

df["Cleaned_Text"] = df["Cleaned_Tokens"].apply(
    lambda tokens: " ".join(tokens)
)

# Display the cleaned text
display(df[["Review", "Cleaned_Text"]])
```

```python

# Create a TF-IDF vectorizer
tfidf_vectorizer = TfidfVectorizer()

# Convert cleaned reviews into numerical features
tfidf_matrix = tfidf_vectorizer.fit_transform(df["Cleaned_Text"])

# Display the TF-IDF values
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

print("TF-IDF Output:")
display(tfidf_df.round(3))
```

```python

print("Vocabulary:")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix Shape:")
print(tfidf_matrix.shape)
```

```python

# Display the most important words in each review

feature_names = tfidf_vectorizer.get_feature_names_out()

for i, review in enumerate(reviews):
    scores = tfidf_matrix[i].toarray().flatten()
    top_indices = scores.argsort()[::-1][:3]

    top_words = [
        (feature_names[index], round(scores[index], 3))
        for index in top_indices
        if scores[index] > 0
    ]

    print(f"Review {i + 1}: {review}")
    print("Top TF-IDF words:", top_words)
    print("-" * 50)
```

```python

# Save the TF-IDF matrix as a CSV file

tfidf_df.to_csv("TF-IDF_Output.csv", index=False)

print("TF-IDF output saved successfully!")
```
