```python
import numpy as np
```

```python
# Creating a 1D array
arr_1d = np.array([10, 20, 30, 40, 50])

print("1D Array:", arr_1d)
print("Shape:", arr_1d.shape)
```

```python
# Creating a 2D array
arr_2d = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("2D Array:")
print(arr_2d)
print("Shape:", arr_2d.shape)
```

```python
# Indexing
print("First element:", arr_1d[0])
print("Third element:", arr_1d[2])

# 2D indexing
print("First row:", arr_2d[0])
print("Element at row 2, column 3:", arr_2d[1, 2])
```

```python
# Slicing
print("First three elements:", arr_1d[:3])
print("Elements from index 2:", arr_1d[2:])
print("2D slice:")
print(arr_2d[:, 1:])
```

```python
# Mathematical operations
print("Addition:", arr_1d + 5)
print("Subtraction:", arr_1d - 5)
print("Multiplication:", arr_1d * 2)
print("Division:", arr_1d / 2)
```

```python
# Aggregation functions
print("Sum:", np.sum(arr_1d))
print("Mean:", np.mean(arr_1d))
print("Maximum:", np.max(arr_1d))
print("Minimum:", np.min(arr_1d))
```

```python
# Python List vs NumPy Array

python_list = [10, 20, 30]
numpy_array = np.array([10, 20, 30])

print("Python List * 2:", python_list * 2)
print("NumPy Array * 2:", numpy_array * 2)
```
