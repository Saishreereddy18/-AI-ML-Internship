```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python

# Load the published UCI Student Performance dataset

df = pd.read_csv("student-mat.csv", sep=";")

print("Real dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()

```

```python

# Check missing values in each column
missing_values = df.isnull().sum()

print("Missing Values Before Cleaning:")
print(missing_values)
```

```python

# Fill missing numerical values with their column mean
numeric_columns = df.select_dtypes(include=np.number).columns

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].mean()
)

# Fill missing categorical values with the most frequent value
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    if df[column].isnull().any():
        df[column] = df[column].fillna(df[column].mode()[0])

print("Missing values handled successfully!")
print(df.isnull().sum())
```

```python

# Remove duplicate rows
df = df.drop_duplicates()

print("Duplicate records removed.")
print("Dataset shape after cleaning:", df.shape)
```

```python

# Calculate each student's average grade

df["Average_Score"] = df[["G1", "G2", "G3"]].mean(axis=1)

print("Average scores calculated successfully!")

df[["G1", "G2", "G3", "Average_Score"]].head()
```

```python

# Categorize students based on average score

def categorize_performance(score):
    if score >= 16:
        return "Excellent"
    elif score >= 12:
        return "Good"
    elif score >= 10:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["Average_Score"].apply(
    categorize_performance
)

print("Performance categories created!")

df[["G1", "G2", "G3", "Average_Score", "Performance"]].head()
```

```python


# Categorize students based on their average grade

def categorize_performance(score):
    if score >= 16:
        return "Excellent"
    elif score >= 12:
        return "Good"
    elif score >= 10:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["Average_Score"].apply(categorize_performance)

print("Performance categories created successfully!")

df[["G1", "G2", "G3", "Average_Score", "Performance"]].head()
```

```python

# Basic analysis of student performance

print("Total students:", len(df))
print("Average final grade (G3):", round(df["G3"].mean(), 2))
print("Highest final grade:", df["G3"].max())
print("Lowest final grade:", df["G3"].min())

print("\nPerformance category counts:")
print(df["Performance"].value_counts())
```

```python

# Identify the student with the highest average score

top_student = df.loc[df["Average_Score"].idxmax()]

print("Top Performing Student")
print("Student ID:", top_student["Student_ID"])
print("Average Score:", round(top_student["Average_Score"], 2))
print("Performance:", top_student["Performance"])
```

```python


# Analyze study time and final grades

study_analysis = df.groupby("studytime")["G3"].mean()

print("Average final grade by weekly study-time category:")
print(study_analysis.round(2))
```

```python

# Count students in each category

performance_counts = df["Performance"].value_counts()

print("Student Performance Distribution:")
print(performance_counts)
```

```python


# Use NumPy to calculate overall score statistics

scores = df["G3"].to_numpy()

print("Mean final grade:", np.mean(scores))
print("Median final grade:", np.median(scores))
print("Highest final grade:", np.max(scores))
print("Lowest final grade:", np.min(scores))
print("Standard deviation:", round(np.std(scores), 2))
```

```python

# Bar chart: Average grades across periods

period_averages = df[["G1", "G2", "G3"]].mean()

plt.figure(figsize=(7, 5))
period_averages.plot(kind="bar")
plt.title("Average Student Grades Across Periods")
plt.xlabel("Grade Period")
plt.ylabel("Average Grade (out of 20)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Scatter plot: Weekly study-time category vs final grade

plt.figure(figsize=(7, 5))
sns.stripplot(data=df, x="studytime", y="G3", jitter=0.2)
plt.title("Weekly Study Time Category vs Final Grade")
plt.xlabel("Study Time Category (1 = lowest, 4 = highest)")
plt.ylabel("Final Grade (G3)")
plt.tight_layout()
plt.show()
```

```python

# Histogram: Distribution of final grades

plt.figure(figsize=(7, 5))
plt.hist(df["G3"], bins=10, edgecolor="black")
plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python

# Pie chart: Performance categories

performance_counts = df["Performance"].value_counts()

plt.figure(figsize=(7, 5))
plt.pie(
    performance_counts,
    labels=performance_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Student Performance Categories")
plt.tight_layout()
plt.show()
```
