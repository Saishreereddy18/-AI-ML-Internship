```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

prompt = "Artificial Intelligence is useful because"

result = generator(
    prompt,
    max_new_tokens=40,
    do_sample=False,
    pad_token_id=generator.tokenizer.eos_token_id
)

generated_text = result[0]["generated_text"]

print("Input Prompt:")
print(prompt)

print("\nGenerated Text:")
print(generated_text)
```

```python
from transformers import pipeline

sentiment_model = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

sentences = [
    "This AI internship is helping me learn new skills.",
    "The application is confusing and frustrating.",
    "The project is useful, but the documentation is poor."
]

sentiment_results = sentiment_model(sentences)

for sentence, result in zip(sentences, sentiment_results):
    print("Text:", sentence)
    print("Sentiment:", result["label"])
    print("Confidence:", round(result["score"], 4))
    print("-" * 50)
```

```python

from transformers import pipeline

summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn"
)

article = """
Artificial Intelligence is changing many industries.
Machine learning identifies patterns in data.
Large Language Models can generate text, answer questions,
summarize documents, and support chatbots.
Hugging Face provides pretrained models that developers
can use in AI applications. AI outputs should be checked
for accuracy, relevance, and bias.
"""

summary = summarizer(
    article,
    max_length=60,
    min_length=20,
    do_sample=False
)

print("Original Article:\n", article)
print("\nGenerated Summary:\n", summary[0]["summary_text"])
```

# Day 15 – Large Language Models (LLMs)

## 1. Objective
Understand Large Language Models and learn how modern AI applications use them.

## 2. Evolution of AI
- **Machine Learning (ML):** Learns patterns from data.
- **Deep Learning (DL):** Uses neural networks to learn complex patterns.
- **Natural Language Processing (NLP):** Helps computers process human language.
- **Generative AI:** Generates text, images, code, and other content.
- **Large Language Models (LLMs):** Models trained on large amounts of text to perform language tasks.

These areas are related and overlapping, rather than strictly separate stages.

## 3. Hugging Face
Hugging Face is a platform for discovering and using pretrained AI models. Its Model Hub provides models for text generation, sentiment analysis, summarization, and other tasks.

Website: https://huggingface.co/models

## 4. Model Comparison
1. **DistilGPT-2:** Generates text from an input prompt.
2. **DistilBERT SST-2:** Classifies text sentiment as positive or negative.
3. **BART Large CNN:** Produces a shorter summary of an article.

Each model performs a different task, so there is no single universal best model.

## 5. Conclusion
This experiment demonstrates how pretrained language models can be used for text generation, sentiment analysis, and summarization. Model selection depends on the task, output quality, accuracy, and available computing resources.

## 6. Results and Comparison

| Model | Task | Observation |
|---|---|---|
| DistilGPT-2 | Text Generation | Produces continuation text from a prompt. |
| DistilBERT SST-2 | Sentiment Analysis | Predicts positive or negative sentiment with a confidence score. |
| BART Large CNN | Summarization | Generates a concise version of the input article. |

**Best model for my chosen task:** BART Large CNN for text summarization, because it is specifically designed to summarize articles. This does not mean it is the best model for every AI task.
