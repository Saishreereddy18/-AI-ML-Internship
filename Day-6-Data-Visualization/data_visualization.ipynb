```python
#Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_theme(style="whitegrid")
os.makedirs("charts", exist_ok=True)

print("Libraries imported successfully!")
```

```python
#Load the Day 5 dataset
df = pd.read_csv("../Day-5-Pandas/student_data.csv")

display(df.head())
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
```

```python
#Check the data
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe())
```

```python
#Bar chart
plt.figure(figsize=(10, 5))

sns.barplot(data=df, x="Name", y="Marks", hue="Name", legend=False)

plt.title("Student Marks Comparison")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("charts/bar_chart.png", dpi=300)
plt.show()
```

```python
#Line chart
plt.figure(figsize=(10, 5))

sns.lineplot(data=df, x="Student_ID", y="Marks", marker="o")

plt.title("Student Marks Trend")
plt.xlabel("Student ID")
plt.ylabel("Marks")
plt.tight_layout()

plt.savefig("charts/line_chart.png", dpi=300)
plt.show()
```

```python
#Histogram
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Marks", bins=5, kde=True)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.tight_layout()

plt.savefig("charts/histogram.png", dpi=300)
plt.show()
```

```python
#Pie chart
gender_counts = df["Gender"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Gender Distribution of Students")
plt.tight_layout()

plt.savefig("charts/pie_chart.png", dpi=300)
plt.show()
```

## Observations from Data Visualization

1. **Marks comparison:** The bar chart identifies the students with the highest and lowest marks.
2. **Marks variation:** The line chart shows how student marks vary across Student IDs.
3. **Marks distribution:** The histogram shows which range of marks contains the most students.
4. **Gender distribution:** The pie chart displays the percentage of students in each gender category.
5. **Overall finding:** The visualizations make it easier to compare student performance and understand the characteristics of the dataset.
