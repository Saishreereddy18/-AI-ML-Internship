```python

import pandas as pd

data = {
    " Student Name ": [
        "Sai", "Ravi", "Anu", "Kiran",
        "Ravi", "Meena", "Arjun", "Divya"
    ],
    "Age ": [
        "20", "21", None, "20",
        "21", "twenty", "22", "19"
    ],
    " Marks ": [
        "85", "90", "78", None,
        "90", "88", None, "92"
    ],
    "Department ": [
        "CSE", "ECE", "CSE", "IT",
        "ECE", "CSE", "ECE", None
    ]
}

df = pd.DataFrame(data)

df.to_csv("student_data_raw.csv", index=False)

print("Sample dataset created successfully!")
print(df)
```

```python

df = pd.read_csv("student_data_raw.csv")

print("First five rows:")
display(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python

print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python

# Remove extra spaces and standardize column names
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

print("Cleaned column names:")
print(df.columns.tolist())

display(df.head())
```

```python

# Convert age and marks to numeric values
df["age"] = pd.to_numeric(df["age"], errors="coerce")
df["marks"] = pd.to_numeric(df["marks"], errors="coerce")

# Fill missing numerical values with the median
df["age"] = df["age"].fillna(df["age"].median())
df["marks"] = df["marks"].fillna(df["marks"].median())

# Fill missing department with the most frequent department
df["department"] = df["department"].fillna(df["department"].mode()[0])

print("Missing values after handling:")
print(df.isnull().sum())

display(df)
```

```python

before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Duplicate rows remaining:", df.duplicated().sum())
```

```python

# Convert age to integer
df["age"] = df["age"].round().astype(int)

# Save the cleaned dataset
df.to_csv("cleaned_student_data.csv", index=False)

print("Cleaned dataset saved successfully!")
print("\nFinal dataset information:")
df.info()

print("\nFinal dataset:")
display(df)

print("\nMissing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

```
