```python
#Filter the data
#Students who scored above 70
df[df["Marks"] > 70]
#Students who scored below 50
df[df["Marks"] < 50]
#Students with attendance above 90
df[df["Attendance"] > 90]
```

```python
#Multiple conditions
df[(df["Marks"] > 70) & (df["Attendance"] > 85)]
```

```python
#Summary Statistics
df.describe()
```

```python
#simple analysis
average_marks = df["Marks"].mean()

print("Average Marks:", average_marks)
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
print("Average Attendance:", df["Attendance"].mean())
```
